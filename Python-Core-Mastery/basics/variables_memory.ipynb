```python
a = 10
b = a
a = 20

print(b)

a = 10        # a points to object 10
b = a         # b points to SAME object 10
a = 20        # a now points to NEW object 20
```

```python
# Memory Identity
a = 10
b = 10

print(a is b)

a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)  # True
print(a is c)  # False
print(a == c)  # True

'''
| Operator | Meaning             |
| -------- | ------------------- |
| `=`      | Assign value        |
| `is`     | Same memory object? |
| `==`     | Same value?         |

'''
```

```python
a = 1000
b = 1000

d=5
e=5

print(a is b)
print(d is e)

'''
Python interns small integers (usually -5 to 256)
👉 Python reuses small integer objects
👉 This is called interning
'''

a = 256
b = 256

print(a is b) # True

a = 257
b = 257

print(a is b) # False
```

```python
# string interning
a = "hello"
b = "hello"

print(a is b)

a = "hello world"
b = "hello world"

print(a is b)

''' 
👉 Only letters, numbers, underscores → more likely interned
Contains space → not always interned

'''
```
